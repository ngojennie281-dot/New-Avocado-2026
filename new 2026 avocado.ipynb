```python
import pandas as pd

# Load data
data2025 = pd.read_csv("2025Data.csv")
data2026 = pd.read_csv("2026Data.csv")

# Keep comparable Jan-Aug Total U.S. data
data2025 = data2025[(data2025["Geography"] == "Total U.S.") & (data2025["Period"] <= 32)]
data2026 = data2026[(data2026["Geography"] == "Total U.S.") & (data2026["Period"] <= 32)]

# Calculate ASP and volume by avocado type
summary2025 = data2025.groupby("Type").agg(
    ASP=("ASP Current Year", "mean"),
    Avg_Volume=("Total Bulk and Bags", "mean"),
    Total_Volume=("Total Bulk and Bags", "sum")
)

summary2026 = data2026.groupby("Type").agg(
    ASP=("ASP Current Year", "mean"),
    Avg_Volume=("Total Bulk and Bags", "mean"),
    Total_Volume=("Total Bulk and Bags", "sum")
)

# Compare 2025 to 2026
comparison = summary2025.add_suffix("_2025").join(summary2026.add_suffix("_2026"))

comparison["ASP_%_Change"] = (
    (comparison["ASP_2026"] / comparison["ASP_2025"]) - 1
) * 100

comparison["Volume_%_Change"] = (
    (comparison["Avg_Volume_2026"] / comparison["Avg_Volume_2025"]) - 1
) * 100

comparison.round(2)
```
